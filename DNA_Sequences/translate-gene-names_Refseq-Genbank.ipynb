# Translate gene names from Refseq to Genbank

This code reads a file and translates the gene names in a target file from Refseq to Genbank. The Refseq-Genbank conversions are based on the GCF_000018685.1_ASM1868v1 GFF file that was read using gene-names_Refseq-Genbank.ipynb. Refseq genes for which there is no Genbank equivalent are left unchanged.


# Setup

```python
# libraries
import sys
```

```python
# File IO

# File containing Refseq->Genbank conversions
refseq_genbank = "/home/tolonen/Drives/Genoscope-googledrive/Lab_Projects/Genome_Cphy/Genome_Files/2025.11_Refseq/gene-names_Refseq_Genbank.tsv"

# file to translate
cog_file = "/home/tolonen/Drives/Genoscope-googledrive/Lab_Projects/Genome_Cphy/COG/cogs_GCF_000018685.1.txt"

# output file
output_file = "/home/tolonen/Drives/Genoscope-googledrive/Lab_Projects/Genome_Cphy/COG/cogs_GCF_000018685.1_translated.txt"
```

# Functions

# Main

```python
# Read Refseq->Genbank conversions into a dictionary

# Read file line-by-line. If the line does not start with "CPHY_RS", skip it. 
# Otherwise, split the line by tabs and store the first column as key and second column as value in a dictionary called gene_names.

gene_names = {}
with open(refseq_genbank, 'r') as file:
    for line in file:
        line = line.strip()
        if not line.startswith("CPHY_RS"):
            continue  
        columns = line.split('\t', 1)
        key = columns[0]   # The first column is the key (e.g., CPHY_RSxxxxx)
        value = columns[1] # The second column is the value (e.g., Gene Name)
        gene_names[key] = value

# Verify
# Use len() to get the count of items
num_pairs = len(gene_names)

# Print the result
print(f"{num_pairs} Refseq gene names were grabbed from the refseq_genbank file.")
            
```

```python
# Translate gene names from Refseq to Genbank in the translate file.

# Read the input_file line-by-line. If the line does not start with "CPHY_RS", then skip it. 
# Otherwise, split the line by tabs and store the first column as key and second column as value in a dictionary called cogs

cogs = {}
with open(cog_file, 'r') as file:
    for line in file:
        line = line.strip()
        if not line.startswith("CPHY_RS"):
            continue  
        columns = line.split('\t', 1)
        key = columns[0]   # The first column is the key (e.g., CPHY_RSxxxxx)
        value = columns[1] # The second column is the value (e.g., Gene Name)
        cogs[key] = value

# Verify
# Use len() to get the count of items
num_pairs = len(cogs)

# Print the result
print(f"{num_pairs} genes were captured from the cog_file.")
```

```python
# For each of the keys in the dictionary "cogs", set the key as the variable "gene_name_test". 
# Read through all the keys in the dictionary "gene_names". If the key from "gene_names" matches "gene_name_test", then set 
# "gene_name_test" to be equal to the corresponding value from "gene_names".  
# After reading through the dictionary "gene_names", then set key-value pair for dictionary "cogs_translated" where key is "gene_name_test" and
# the value is the corresponding value from "cogs". 

cogs_translated = {}

my_counter = 0
for cogs_key, cogs_value in cogs.items():
    gene_name_test = cogs_key
    if gene_name_test in gene_names:
        gene_name_test = gene_names[gene_name_test]
        my_counter+=1
    cogs_translated[gene_name_test] = cogs_value

print(f"Translated gene names from Refseq to Genbank for {my_counter} genes.")

# Verify dictionary
num_pairs = len(cogs_translated)
print(f"Built dictiony of COGS for {num_pairs} genes where gene names have been replaced with Genbank names, where possible.")

# print output
with open(output_file, 'w') as my_file:
    print(f"List of COGs from cogs_GCF_000018685.1.txt in which the gene names have been translated from Refseq to Genbank.\n", file = my_file)
    for key, value in cogs_translated.items():
        my_file.write(f"{key}\t{value}\n")  
```
