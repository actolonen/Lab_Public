# OpenAI API requests

Code to interact with the OpenAI API

Advantages of interacting with OpenAI using the API
1. Customize system message
2. Adjust input parameters: number of responses, response length, temperature (degree of randomness) of responses
3. Process images
4. Extract word embedding
5. Analyze audio files
6. Fine-tune the model

Some OpenAI models
1. GPT-3.5-turbo (free model)
2. Whisper (speech recognition)
3. Embeddings (text similarity)
4. Dall-E (image creation)
