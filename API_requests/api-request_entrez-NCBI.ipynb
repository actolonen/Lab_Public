# NCBI API queries

Entrez = NCBI data retrieval system.

## Modules
`Bio.Entrez` module to access Entrez. Here, Entrez is a module in the Biopython library.

## Bio.Entrez functions
einfo() get list of dbs
read() parse handle into python dictionary
esearch() search a db for a term
efetch() fetch records from a db. need to specify return type (rettype) and 'retmode' arguments.


```python
from Bio import Entrez, SeqIO # import modules from Bio library
```

```python
Entrez.email = "atolonen@genoscope.cns.fr"
```

```python
# get list of available databases

handle = Entrez.einfo() # einfo is function to get list of databases. handle is xml object
rec = Entrez.read(handle) # parses the xml object into a python dictionary
handle.close()
print(rec.keys()) # pint keys of rec dictionary

# print out key-value pairs of rec dictionary
for key, value in rec.items():
    print(f"key: {key}\tvalue: {value}")
```

```python
# query nucleotide database.
handle = Entrez.esearch(db = "nucleotide", term = "cphy_3367[Gene Name]")
rec_list = Entrez.read(handle)
handle.close()
print(rec_list["IdList"]) # print list of IDs matching the query
            
```

```python
# get sequence associated with records grabbed with esearch
id_list = rec_list["IdList"] # access values from key "IdList"
handle = Entrez.efetch(db="nucleotide", id=id_list, rettype="gb")

recs = list(SeqIO.parse(handle, "gb")) # parse the handle into a list of SeqRecord objects
handle.close()
```

```python
for rec in recs:
    if rec.name == "CP000885":
        print(rec.name)
        print(rec.description)
```
